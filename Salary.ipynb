```python
def salary(x):
    if x>=1 and x<=400:
        return 'Mid_n'
    elif x>400 and x<=800:
        return 'early_m'
    elif x>800 and x<=1200:
        return 'morning'
    elif x>1200 and x<=1600:
        return 'afternoon'
    elif x>1600 and x<=2000:
        return 'evening'
    else:
        return 'night'
    
```

```python
df['Salary']=df.YearsExperience.apply(salary)
df
```
